## **Experiment 5.1:** Implement subword-level tokenization using a BPE-based tokenizer. Given input data:
`https://drive.google.com/file/d/1eHA0Q9ju-08n_hByCK03GqWAX9HQvAKd/view?usp=sharing`

## Tokenize the sentence into subword units.

## Display the generated tokens and their corresponding token IDs.
(Implement two separate codes: a. Using pretrained model and b. Without using pretrained model)


## a. Using Pretrained Model

````python
from transformers import GPT2TokenizerFast

# Load pretrained GPT-2 BPE tokenizer
tokenizer = GPT2TokenizerFast.from_pretrained("gpt2")

# Read the given input data
file_path = "input_sub_word_data.txt"

with open(file_path, "r", encoding="utf-8") as file:
    text = file.read()

# Take the first sentence from the input data
sentence = text.split(".")[0].strip() + "."

print("Input Sentence:")
print(sentence)

# Tokenize sentence into subword units
tokens = tokenizer.tokenize(sentence)

# Convert tokens to token IDs
token_ids = tokenizer.convert_tokens_to_ids(tokens)

print("\nGenerated BPE Tokens and Token IDs:")
print("-" * 45)

for token, token_id in zip(tokens, token_ids):
    print(f"{token:<20} -> {token_id}")

print("\nTokens:")
print(tokens)

print("\nToken IDs:")
print(token_ids)
````

## b. Without Using Pretrained Model

````python
from tokenizers import Tokenizer
from tokenizers.models import BPE
from tokenizers.trainers import BpeTrainer
from tokenizers.pre_tokenizers import Whitespace

# -----------------------------------
# 1. Create BPE tokenizer from scratch
# -----------------------------------

tokenizer = Tokenizer(
    BPE(unk_token="[UNK]")
)

# Split text into words before applying BPE
tokenizer.pre_tokenizer = Whitespace()

# -----------------------------------
# 2. Configure BPE trainer
# -----------------------------------

trainer = BpeTrainer(
    vocab_size=500,
    min_frequency=2,
    special_tokens=["[UNK]"]
)

# -----------------------------------
# 3. Train using the given input file
# -----------------------------------

file_path = "input_sub_word_data.txt"

tokenizer.train(
    files=[file_path],
    trainer=trainer
)

# -----------------------------------
# 4. Read input data
# -----------------------------------

with open(file_path, "r", encoding="utf-8") as file:
    text = file.read()

# Take first sentence for testing
sentence = text.split(".")[0].strip() + "."

print("Input Sentence:")
print(sentence)

# -----------------------------------
# 5. Apply trained BPE tokenizer
# -----------------------------------

encoding = tokenizer.encode(sentence)

tokens = encoding.tokens
token_ids = encoding.ids

# -----------------------------------
# 6. Display tokens and IDs
# -----------------------------------

print("\nGenerated BPE Tokens and Token IDs:")
print("-" * 45)

for token, token_id in zip(tokens, token_ids):
    print(f"{token:<20} -> {token_id}")

print("\nTokens:")
print(tokens)

print("\nToken IDs:")
print(token_ids)
````

## **Experiment 5.2:** Implement subword-level tokenization using a SentencePiece tokenizer. Given same input data of Experiment 5.1
`https://drive.google.com/file/d/1eHA0Q9ju-08n_hByCK03GqWAX9HQvAKd/view?usp=sharing`

## Tokenize the sentence into subword units.
## Display the generated tokens and their corresponding token IDs.


## a. Using Pretrained Model

````python
from transformers import AutoTokenizer

# -----------------------------------------
# 1. Load pretrained SentencePiece tokenizer
# -----------------------------------------

tokenizer = AutoTokenizer.from_pretrained(
    "google-t5/t5-small",
    use_fast=False
)

# -----------------------------------------
# 2. Read the given input data
# -----------------------------------------

file_path = "input_sub_word_data.txt"

with open(file_path, "r", encoding="utf-8") as file:
    text = file.read()

# Take the first sentence from the dataset
sentence = text.split(".")[0].strip() + "."

print("Input Sentence:")
print(sentence)

# -----------------------------------------
# 3. Tokenize into subword units
# -----------------------------------------

tokens = tokenizer.tokenize(sentence)

# -----------------------------------------
# 4. Convert tokens into token IDs
# -----------------------------------------

token_ids = tokenizer.convert_tokens_to_ids(tokens)

# -----------------------------------------
# 5. Display tokens and token IDs
# -----------------------------------------

print("\nSentencePiece Tokens and Token IDs:")
print("-" * 50)

for token, token_id in zip(tokens, token_ids):
    print(f"{token:<25} -> {token_id}")

print("\nGenerated Tokens:")
print(tokens)

print("\nToken IDs:")
print(token_ids)
````

## b. Without Using Pretrained Model

````python
# Install required library
# pip install sentencepiece

import sentencepiece as spm

# -----------------------------------------
# 1. Specify the given input corpus
# -----------------------------------------

input_file = "input_sub_word_data.txt"

# -----------------------------------------
# 2. Train SentencePiece from scratch
# -----------------------------------------

spm.SentencePieceTrainer.train(
    input=input_file,
    model_prefix="my_sentencepiece",
    vocab_size=300,
    model_type="bpe",
    character_coverage=1.0
)

print("SentencePiece model trained successfully!")

# -----------------------------------------
# 3. Load our newly trained model
# -----------------------------------------

sp = spm.SentencePieceProcessor()

sp.load("my_sentencepiece.model")

# -----------------------------------------
# 4. Read the given dataset
# -----------------------------------------

with open(input_file, "r", encoding="utf-8") as file:
    text = file.read()

# Take first sentence for testing
sentence = text.split(".")[0].strip() + "."

print("\nInput Sentence:")
print(sentence)

# -----------------------------------------
# 5. Tokenize into subword units
# -----------------------------------------

tokens = sp.encode(
    sentence,
    out_type=str
)

# -----------------------------------------
# 6. Generate corresponding token IDs
# -----------------------------------------

token_ids = sp.encode(
    sentence,
    out_type=int
)

# -----------------------------------------
# 7. Display tokens and token IDs
# -----------------------------------------

print("\nSentencePiece Tokens and Token IDs:")
print("-" * 50)

for token, token_id in zip(tokens, token_ids):
    print(f"{token:<25} -> {token_id}")

print("\nGenerated Tokens:")
print(tokens)

print("\nToken IDs:")
print(token_ids)
````

## **Experiment 5.3:** Implement Part-of-Speech (POS) tagging on a given text and display the unique token, POS tag, and description of each POS tag.

`"The young student is reading an interesting book in the library."`

## a. Using spaCy

````python
import spacy

# Load the English language model
nlp = spacy.load("en_core_web_sm")

# Given text
text = "The young student is reading an interesting book in the library."

# Process the text
doc = nlp(text)

# Set to store unique tokens
seen = set()

print(f"{'Token':<15} {'POS Tag':<12} {'Description'}")
print("-" * 60)

# Display unique token, POS tag and description
for token in doc:

    # Convert to lowercase for checking uniqueness
    key = token.text.lower()

    if key not in seen:
        seen.add(key)

        # Get POS description
        description = spacy.explain(token.pos_)

        print(
            f"{token.text:<15} "
            f"{token.pos_:<12} "
            f"{description}"
        )
````

## b. Using NLTK

````python
import nltk

# Download required NLTK resources
nltk.download("punkt")
nltk.download("punkt_tab")
nltk.download("averaged_perceptron_tagger")
nltk.download("averaged_perceptron_tagger_eng")
````

````python
# Given text
text = "The young student is reading an interesting book in the library."

# Tokenize the sentence
tokens = nltk.word_tokenize(text)

# Perform POS tagging
tagged_tokens = nltk.pos_tag(tokens)

# Dictionary containing descriptions of common Penn Treebank POS tags
tag_descriptions = {
    "CC": "Coordinating conjunction",
    "CD": "Cardinal number",
    "DT": "Determiner",
    "EX": "Existential there",
    "FW": "Foreign word",
    "IN": "Preposition or subordinating conjunction",
    "JJ": "Adjective",
    "JJR": "Adjective, comparative",
    "JJS": "Adjective, superlative",
    "MD": "Modal",
    "NN": "Noun, singular or mass",
    "NNS": "Noun, plural",
    "NNP": "Proper noun, singular",
    "NNPS": "Proper noun, plural",
    "PDT": "Predeterminer",
    "POS": "Possessive ending",
    "PRP": "Personal pronoun",
    "PRP$": "Possessive pronoun",
    "RB": "Adverb",
    "RBR": "Adverb, comparative",
    "RBS": "Adverb, superlative",
    "RP": "Particle",
    "TO": "to",
    "UH": "Interjection",
    "VB": "Verb, base form",
    "VBD": "Verb, past tense",
    "VBG": "Verb, gerund/present participle",
    "VBN": "Verb, past participle",
    "VBP": "Verb, non-3rd person singular present",
    "VBZ": "Verb, 3rd person singular present",
    "WDT": "Wh-determiner",
    "WP": "Wh-pronoun",
    "WP$": "Possessive wh-pronoun",
    "WRB": "Wh-adverb"
}

# Set for storing unique tokens
seen = set()

print(f"{'Token':<15} {'POS Tag':<10} {'Description'}")
print("-" * 65)

# Display unique tokens
for token, tag in tagged_tokens:

    key = token.lower()

    if key not in seen:
        seen.add(key)

        description = tag_descriptions.get(
            tag,
            "Punctuation or other tag"
        )

        print(
            f"{token:<15} "
            f"{tag:<10} "
            f"{description}"
        )
````

## **Experiment 5.4:** Implement Part-of-Speech (POS) tagging on a given text and display the unique token, POS tag, description of each token and frequency using Spacy.

## Input data: Same input data of Experiment 5.1

`https://drive.google.com/file/d/1eHA0Q9ju-08n_hByCK03GqWAX9HQvAKd/view?usp=sharing`


````python
import spacy
from collections import Counter

# -----------------------------------------
# 1. Load the spaCy English model
# -----------------------------------------

nlp = spacy.load("en_core_web_sm")

# -----------------------------------------
# 2. Read the given input data
# -----------------------------------------

file_path = "input_sub_word_data.txt"

with open(file_path, "r", encoding="utf-8") as file:
    text = file.read()

# -----------------------------------------
# 3. Process the text using spaCy
# -----------------------------------------

doc = nlp(text)

# -----------------------------------------
# 4. Calculate frequency of each token
# -----------------------------------------

# Convert tokens to lowercase so that
# "Natural" and "natural" are treated as
# the same token.

frequency = Counter(
    token.text.lower()
    for token in doc
    if not token.is_space
)

# -----------------------------------------
# 5. Display table heading
# -----------------------------------------

print(
    f"{'Unique Token':<20}"
    f"{'POS Tag':<12}"
    f"{'Description':<30}"
    f"{'Frequency'}"
)

print("-" * 75)

# -----------------------------------------
# 6. Store tokens already displayed
# -----------------------------------------

seen = set()

# -----------------------------------------
# 7. Display unique tokens
# -----------------------------------------

for token in doc:

    # Ignore whitespace
    if token.is_space:
        continue

    # Convert token to lowercase
    word = token.text.lower()

    # Display each token only once
    if word not in seen:

        seen.add(word)

        # Get POS tag
        pos_tag = token.pos_

        # Get description of POS tag
        description = spacy.explain(pos_tag)

        # Get frequency
        freq = frequency[word]

        print(
            f"{word:<20}"
            f"{pos_tag:<12}"
            f"{description:<30}"
            f"{freq}"
        )
````

## **Experiment 5.5:** Write a Python program to train a Word2Vec CBOW model using the given text corpus. Display the vector representation of a selected word and find the three most similar words.

## **Input corpus**

### Create a file named corpus.txt:

```
Natural language processing is interesting

Natural language processing is useful

I love machine learning

Machine learning is useful

Deep learning is a part of machine learning

Artificial intelligence is changing the world

Natural language processing uses machine learning

Deep learning and artificial intelligence are related

I study natural language processing

Machine learning helps artificial intelligence
```

````python
from gensim.models import Word2Vec

# Read corpus
with open("corpus.txt", "r", encoding="utf-8") as file:
    sentences = [
        line.lower().strip().split()
        for line in file
        if line.strip()
    ]

# Train CBOW model
# sg=0 means CBOW
model = Word2Vec(
    sentences=sentences,
    vector_size=50,
    window=3,
    min_count=1,
    workers=1,
    sg=0,
    epochs=100,
    seed=42
)

selected_word = "machine"

# Display vector
print(f"Vector representation of '{selected_word}':")
print(model.wv[selected_word])

# Three most similar words
print(f"\nThree words most similar to '{selected_word}':")

similar_words = model.wv.most_similar(
    selected_word,
    topn=3
)

for word, similarity in similar_words:
    print(f"{word:15} {similarity:.4f}")
````

## **Experiment 5.6:** Write a Python program to train a Skip-Gram Word2Vec model using the same corpus. Retrieve the vector for "language" and display the five most similar words.

### **Input corpus:** Use corpus.txt from Question 5.5.

````python
from gensim.models import Word2Vec

# Read corpus
with open("corpus.txt", "r", encoding="utf-8") as file:
    sentences = [
        line.lower().strip().split()
        for line in file
        if line.strip()
    ]

# Train Skip-Gram Word2Vec
# sg=1 means Skip-Gram
model = Word2Vec(
    sentences=sentences,
    vector_size=50,
    window=3,
    min_count=1,
    workers=1,
    sg=1,
    epochs=100,
    seed=42
)

word = "language"

# Retrieve vector
vector = model.wv[word]

print(f"Vector representation of '{word}':")
print(vector)

# Find five most similar words
print(f"\nFive most similar words to '{word}':")

similar_words = model.wv.most_similar(
    word,
    topn=5
)

for similar_word, similarity in similar_words:
    print(
        f"{similar_word:15} "
        f"{similarity:.4f}"
    )
````

## **Experiment 5.7:** Train a Word2Vec model using the provided corpus and calculate the cosine similarity between selected word pairs.

### **Input corpus:** Use corpus.txt from Question 5.5.

### Calculate similarity for:
### a. `machine and learning`
### b. `natural and language`
### c. `deep and learning`

````python
from gensim.models import Word2Vec

# Read corpus
with open("corpus.txt", "r", encoding="utf-8") as file:
    sentences = [
        line.lower().strip().split()
        for line in file
        if line.strip()
    ]

# Train Word2Vec model
model = Word2Vec(
    sentences=sentences,
    vector_size=50,
    window=3,
    min_count=1,
    workers=1,
    sg=1,
    epochs=100,
    seed=42
)

# Required word pairs
word_pairs = [
    ("machine", "learning"),
    ("natural", "language"),
    ("deep", "learning")
]

print("Cosine Similarities")
print("-" * 45)

for word1, word2 in word_pairs:

    similarity = model.wv.similarity(
        word1,
        word2
    )

    print(
        f"{word1:10} - "
        f"{word2:10}: "
        f"{similarity:.4f}"
    )
````
