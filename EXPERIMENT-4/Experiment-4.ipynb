## **Experiment 4.1:** To perform Term-Frequency Analysis and Named Entity Recognition on a given text using any toolkit.

- Calculate the frequency of each word and save the output in CSV format (Term, Frequency)
- Identify the top 10 most frequent terms and display their corresponding frequencies.

```python
import spacy
import pandas as pd
from collections import Counter

# Load spaCy English model
nlp = spacy.load("en_core_web_sm")

# Read input file
with open("4.1_4.2_input.txt", "r", encoding="utf-8") as file:
    text = file.read()

# Process text using spaCy
doc = nlp(text)

# Extract words:
# lowercase + alphabetic tokens only
words = [token.text.lower() for token in doc if token.is_alpha]

# Calculate term frequency
frequency = Counter(words)

# Convert to DataFrame
df = pd.DataFrame(
    frequency.items(),
    columns=["Term", "Frequency"]
)

# Sort by frequency
df = df.sort_values(
    by=["Frequency", "Term"],
    ascending=[False, True]
)

# Save all frequencies
df.to_csv("term_frequency_4_1.csv", index=False)

# Display top 10 terms
print("Top 10 Most Frequent Terms")
print("-" * 30)

print(df.head(10).to_string(index=False))

# Named Entity Recognition (NER)
print("-" * 30)
print('\nNamed Entities found (Top 10):')
print("-" * 30)

# Extract named entities
entities = [(ent.text, ent.label_) for ent in doc.ents]
df_ner = pd.DataFrame(entities, columns=['Entity', 'Label'])
display(df_ner.head(10))
```

## **Experiment 4.2:** To perform Term-Frequency Analysis on a given text without using any toolkit.

- Calculate the frequency of each word and save the output in CSV format (Term, Frequency)
- Identify the top 10 most frequent terms and display their corresponding frequencies.


```python
import string
import csv

# Read input data
with open('4.1_4.2_input.txt', 'r', encoding='utf-8') as f:
    text = f.read()

# Preprocessing: lowercase and remove punctuation
text_lower = text.lower()
translator = str.maketrans('', '', string.punctuation)
clean_text = text_lower.translate(translator)

# Tokenize by splitting on whitespace
tokens = clean_text.split()

# Calculate Term Frequency natively
native_word_freq = {}
for token in tokens:
    native_word_freq[token] = native_word_freq.get(token, 0) + 1

# Identify top 10 most frequent terms
sorted_freq = sorted(native_word_freq.items(), key=lambda x: x[1], reverse=True)
top_10_native = sorted_freq[:10]

# Save output in CSV format
with open(
    "term_frequency_4_2.csv",
    "w",
    newline="",
    encoding="utf-8"
) as file:

    writer = csv.writer(file)

    # Header
    writer.writerow(["Term", "Frequency"])

    # All terms
    writer.writerows(sorted_freq)

# Display top 10
print("Top 10 Most Frequent Terms")
print("-" * 35)
print(f"{'Term':<20} {'Frequency'}")
print("-" * 35)

for term, count in sorted_freq[:10]:
    print(f"{term:<20} {count}")

print("\nCSV file saved as term_frequency_4_2.csv")
```

## **Experiment 4.3:** Write a Python program without using NLTK, spaCy, or any other NLP toolkit to:

### **Given Input**
Consider the following three documents:

**Document 1:**

Natural language processing is a field of artificial intelligence.

**Document 2:**

Natural language processing helps computers understand human language.

**Document 3:**

Machine learning is an important part of artificial intelligence.


```python
import math
import string
```

### **4.3.1** Read multiple text documents.

```python
documents = [
    "Natural language processing is a field of artificial intelligence.",
    "Natural language processing helps computers understand human language.",
    "Machine learning is an important part of artificial intelligence."
]
```

### **4.3.2** Convert text to lowercase.
### **4.3.3** Remove punctuation.
### **4.3.4** Tokenize the documents.

```python
def preprocess(doc):
    doc = doc.lower()
    doc = doc.translate(str.maketrans('', '', string.punctuation))
    return doc.split()

tokenized_docs = [preprocess(doc) for doc in documents]


print("PREPROCESSED DOCUMENTS")
print("=" * 70)

for i, tokens in enumerate(tokenized_docs, start=1):
    print("Document", i, ":", tokens)
```

### **4.3.5** Calculate TF for every term.

```python
tf_docs = []

for tokens in tokenized_docs:

    tf = {}

    total_words = len(tokens)

    for word in tokens:
        tf[word] = tf.get(word, 0) + 1

    for word in tf:
        tf[word] = tf[word] / total_words  # Term Frequency formulation

    tf_docs.append(tf)
```

### **4.3.6** Calculate DF for every term.

```python
df = {}

for tokens in tokenized_docs:
    unique_terms = set(tokens)

    for term in unique_terms:
        df[term] = df.get(term, 0) + 1
```

### **4.3.7** Calculate IDF using the logarithmic formula.

```python
total_docs = len(documents)
idf = {}

for term, freq in df.items():
    idf[term] = math.log(total_docs / freq)
```

### **4.3.8** Calculate TF-IDF for every term in every document.

```python
tfidf_docs = []

for tf in tf_docs:
    tfidf = {}

    for term, tf_value in tf.items():
        tfidf[term] = tf_value * idf[term]

    tfidf_docs.append(tfidf)
```

### **4.3.9** Display Term, TF, DF, IDF, and TF-IDF at each step.

```python
for i, doc_tfidf in enumerate(tfidf_docs):

    print(f"\n--- Document {i + 1} ---")

    print(
        f'{"Term":<15} | '
        f'{"TF":<8} | '
        f'{"DF":<4} | '
        f'{"IDF":<8} | '
        f'{"TF-IDF"}'
    )

    print("-" * 60)

    displayed = set()

    for term in tokenized_docs[i]:

        if term not in displayed:

            t_tf = tf_docs[i][term]
            t_df = df[term]
            t_idf = idf[term]
            t_tfidf = doc_tfidf[term]

            print(
                f"{term:<15} | "
                f"{t_tf:<8.4f} | "
                f"{t_df:<4} | "
                f"{t_idf:<8.4f} | "
                f"{t_tfidf:.4f}"
            )

            displayed.add(term)
```

### **4.3.10** Identify the top 10 terms with the highest TF-IDF score for each document.

```python
top_10 = sorted(
    doc_tfidf.items(),
    key=lambda x: x[1],
    reverse=True)[:10]

print("\nTop 10 Terms (Highest TF-IDF):")
print("-" * 40)

for rank, (term, score) in enumerate(top_10, start=1):
  print(f"{rank}. {term}: {score:.4f}")
```
