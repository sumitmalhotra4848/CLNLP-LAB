# <u>**EXPERIMENT-1: DATA ANALYSIS**</u>

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

## 1. Read the employee dataset and display the average salary of each department using a bar chart.

```python
# Read dataset
df = pd.read_csv("employee_information_100.csv")
df.head()
```

```python
avg_salary = df.groupby("Department")["Salary"].mean()

plt.figure(figsize=(8,5))
avg_salary.plot(kind="bar", color="skyblue")
plt.title("Average Salary by Department")
plt.xlabel("Department")
plt.ylabel("Average Salary")
plt.xticks(rotation=45)
plt.show()
```

## 2. Display the number of employees working in each department.

```python
dept_count = df["Department"].value_counts()
print("Employee counts per department:")
print(dept_count)

plt.figure(figsize=(8,5))
dept_count.plot(kind="bar", color="orange")
plt.title("Employees in Each Department")
plt.xlabel("Department")
plt.ylabel("Number of Employees")
plt.xticks(rotation=45)
plt.show()
```

## 3. Display the percentage of male and female employees using a pie chart.

```python
gender_count = df["Gender"].value_counts()

plt.figure(figsize=(6,6))
plt.pie(gender_count,
        labels=gender_count.index,
        autopct="%1.1f%%",
        colors=['lightblue', 'pink'],
        startangle=90)
plt.title("Gender Distribution")
plt.show()
```

## 4. Draw a histogram showing salary distribution.

```python
plt.figure(figsize=(8,5))
sns.histplot(df['Salary'], bins=15, kde=True, label="Salary")
plt.title("Salary Distribution")
plt.xlabel("Salary")
plt.ylabel("Frequency")
plt.legend()
plt.show()
```

## 5. Draw a scatter plot showing the relationship between experience and salary.

```python
plt.figure(figsize=(8,5))
sns.scatterplot(data=df, x='Experience_Years', y='Salary', hue='Department', style='Gender')
plt.title("Experience vs Salary")
plt.xlabel("Experience (Years)")
plt.ylabel("Salary")
plt.legend(loc='upper left')
plt.show()
```

## 6. Display the top 10 employees with the highest salary.

```python
top10 = df.nlargest(10, "Salary")

print("\nTop 10 Highest Paid Employees")
top10[["Employee_ID", "Name", "Department", "Salary"]]
```

## 7. Find the highest salary in every department.

```python
highest_salary = df.groupby("Department")["Salary"].max()

print("\nHighest Salary in Each Department")
highest_salary
```

## 8. Display employees whose salary is greater than the overall average salary.

```python
overall_avg = df["Salary"].mean()

above_avg = df[df["Salary"] > overall_avg]

print("\nOverall Average Salary:", overall_avg)
print("\nEmployees Earning Above Average Salary")
above_avg[["Employee_ID", "Name", "Department", "Salary"]]
```

## 9. Calculate the average years of experience in each department.

```python
avg_exp = df.groupby("Department")["Experience_Years"].mean()

print("\nAverage Experience in Each Department")
avg_exp
```

## 10. Display the distribution of employee ages using a histogram.

```python
plt.figure(figsize=(8,5))
sns.histplot(df['Age'], bins=15, kde=True)
plt.title("Age Distribution")
plt.xlabel("Age")
plt.ylabel("Frequency")
plt.show()
```
