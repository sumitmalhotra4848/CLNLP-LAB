## **Experiment 2.1:**  Write a Python program to perform basic text preprocessing on a given data. The program should:








```python
import string
import re

# Read the text file
with open("2.1_text_data.txt", "r") as file:
    text = file.read()

print("----- ORIGINAL TEXT -----")
print(text)
```

### 1. Convert text to lowercase. (Also display how many uppercase?)

```python
# 1. Convert text to lowercase
uppercase_count = sum(1 for ch in text if ch.isupper())
text = text.lower()

print("\nNumber of uppercase letters:", uppercase_count)
```

### 2. Remove punctuation.        (Also display how many different types of punctuations?)

```python
# 2. Remove punctuation
punctuations = [ch for ch in text if ch in string.punctuation]
different_punctuations = set(punctuations)

print("Number of punctuation marks:", len(punctuations))
print("Number of different types of punctuation:",
      len(different_punctuations))
print("Different punctuations:", different_punctuations)

text = text.translate(str.maketrans("", "", string.punctuation))
```

### 3. Remove numbers.            (Also display how many numbers?)

```python
# 3. Remove numbers
numbers = [ch for ch in text if ch.isdigit()]

print("\nNumber of digits:", len(numbers))

text = re.sub(r'\d+', '', text)
```

### 4. Remove extra whitespace. (How much extra whitespace is present?)

```python
# 4. Remove extra whitespace
# Count whitespace characters beyond one in each consecutive whitespace group
whitespace_groups = re.findall(r'\s+', text)
extra_whitespace = sum(len(group) - 1 for group in whitespace_groups
                       if len(group) > 1)

print("Amount of extra whitespace:", extra_whitespace)

text = re.sub(r'\s+', ' ', text).strip()
```

### 5. Display the cleaned text.

```python
# 5. Display cleaned text
print("\n----- CLEANED TEXT -----")
print(text)
```

## **Experiment 2.2:** Write a Python program to perform word tokenization and sentence tokenization on a given text. Display the resulting words and sentences separately

```python
# Read text from file
with open("2.2_tokenization_data.txt", "r") as file:
    text = file.read()

print("----- ORIGINAL TEXT -----")
print(text)
```

### **1. Using NLTK**

```python
import nltk

# Download required resources (run once)
nltk.download('punkt')
nltk.download('punkt_tab')

# Sentence Tokenization
sentences = nltk.sent_tokenize(text)

print("----- SENTENCE TOKENIZATION -----")
for i, sentence in enumerate(sentences, 1):
    print(i, ":", sentence)

# Word Tokenization
words = nltk.word_tokenize(text)

print("\n----- WORD TOKENIZATION -----")
print(words)
```

### **2. Using spaCy**

```python
import spacy

# Load English language model
nlp = spacy.load("en_core_web_sm")

# Process the text
doc = nlp(text)

# Sentence Tokenization
print("----- SENTENCE TOKENIZATION -----")
for i, sentence in enumerate(doc.sents, 1):
    print(i, ":", sentence.text)

# Word Tokenization
print("\n----- WORD TOKENIZATION -----")
words = [token.text for token in doc]

print(words)
```

### **3. Without NLTK or spaCy**

```python
# Sentence Tokenization
sentences = re.split(r'(?<=[.!?])\s+', text.strip())

print("----- SENTENCE TOKENIZATION -----")
for i, sentence in enumerate(sentences, 1):
    print(i, ":", sentence)

# Word Tokenization
words = re.findall(r"\b\w+(?:[-']\w+)*\b", text)

print("\n----- WORD TOKENIZATION -----")
print(words)
```

# **Experiment 2.3:** Write a Python program to tokenize a text and remove commonly occurring stop words such as is, the, a, an, in, of, and, to, etc. Display the original and filtered tokens. (Also, display Stop Words Separately).

```python
# Read text from file
with open("2.3_clean_data.txt", "r") as file:
    text = file.read()

print("----- ORIGINAL TEXT -----")
print(text)
```

```python
# Define commonly occurring stop words
stop_words = {
    "is", "the", "a", "an", "in", "of", "and", "to",
    "for", "on", "with", "from", "by", "this", "that", "it"
}

# Tokenize the text into words
tokens = re.findall(r"\b\w+\b", text.lower())

# Display original tokens
print("----- ORIGINAL TOKENS -----")
print(tokens)

# Find stop words present in the text
removed_stop_words = [word for word in tokens if word in stop_words]

# Remove stop words
filtered_tokens = [word for word in tokens if word not in stop_words]

# Display stop words separately
print("\n----- STOP WORDS -----")
print(removed_stop_words)

# Display filtered tokens
print("\n----- FILTERED TOKENS -----")
print(filtered_tokens)
```
